```python


import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from datetime import datetime, timedelta
import pathway as pw
import warnings
warnings.filterwarnings('ignore')

def robust_data_preprocessing(csv_path):
    """
    Robust preprocessing that handles various dataset formats
    """
    try:

        df = pd.read_csv(csv_path)
        print(f"Successfully loaded dataset: {df.shape}")

    except FileNotFoundError:
        print(f"File not found: {csv_path}")
        print("Creating sample dataset for demonstration...")


        dates = pd.date_range(start='2024-01-01', periods=1000, freq='30T')
        sample_data = []

        for date in dates:
            for location in range(1, 6):
                capacity = np.random.choice([50, 75, 100, 125, 150])


                hour = date.hour
                day_of_week = date.weekday()

                if day_of_week < 5:  # Weekday
                    if 7 <= hour <= 9 or 17 <= hour <= 19:
                        base_occupancy = 0.8
                    elif 10 <= hour <= 16:
                        base_occupancy = 0.6
                    else:  # Off hours
                        base_occupancy = 0.2
                else:  # Weekend
                    if 10 <= hour <= 16:
                        base_occupancy = 0.7
                    else:
                        base_occupancy = 0.3


                occupancy_rate = max(0, min(1, base_occupancy + np.random.normal(0, 0.15)))
                occupancy = int(occupancy_rate * capacity)

                sample_data.append({
                    'Timestamp': date,
                    'LocationID': location,
                    'Occupancy': occupancy,
                    'Capacity': capacity
                })

        df = pd.DataFrame(sample_data)
        print("Created sample dataset with realistic parking patterns")

    except Exception as e:
        print(f"Error reading file: {e}")
        return None


    print(f"\nOriginal columns: {list(df.columns)}")


    if 'Timestamp' not in df.columns:
        timestamp_candidates = [col for col in df.columns if 'time' in col.lower() or 'date' in col.lower()]

        if timestamp_candidates:
            print(f"Found timestamp candidates: {timestamp_candidates}")

            try:
                df['Timestamp'] = pd.to_datetime(df[timestamp_candidates[0]], infer_datetime_format=True)
                print(f"Successfully created Timestamp from {timestamp_candidates[0]}")
            except:
                print("Could not parse timestamp, creating synthetic timestamps")
                df['Timestamp'] = pd.date_range(start='2024-01-01', periods=len(df), freq='30T')
        else:
            print("No timestamp columns found, creating synthetic timestamps")
            df['Timestamp'] = pd.date_range(start='2024-01-01', periods=len(df), freq='30T')


    if 'Occupancy' not in df.columns:
        occupancy_candidates = [col for col in df.columns if 'occup' in col.lower() or 'used' in col.lower()]

        if occupancy_candidates:
            df['Occupancy'] = df[occupancy_candidates[0]]
            print(f"Using {occupancy_candidates[0]} as Occupancy")
        else:
            print("No occupancy column found, creating synthetic data")
            df['Occupancy'] = np.random.randint(0, 101, len(df))


    if 'Capacity' not in df.columns:
        capacity_candidates = [col for col in df.columns if 'capacity' in col.lower() or 'total' in col.lower()]

        if capacity_candidates:
            df['Capacity'] = df[capacity_candidates[0]]
            print(f"Using {capacity_candidates[0]} as Capacity")
        else:
            print("No capacity column found, creating synthetic data")
            df['Capacity'] = np.random.randint(50, 201, len(df))


    if 'LocationID' not in df.columns:
        location_candidates = [col for col in df.columns if 'location' in col.lower() or 'id' in col.lower()]

        if location_candidates:
            df['LocationID'] = df[location_candidates[0]]
            print(f"Using {location_candidates[0]} as LocationID")
        else:
            print("No location column found, creating synthetic location IDs")
            df['LocationID'] = np.random.randint(1, 6, len(df))


    df['Occupancy'] = pd.to_numeric(df['Occupancy'], errors='coerce').fillna(0).astype(int)
    df['Capacity'] = pd.to_numeric(df['Capacity'], errors='coerce').fillna(100).astype(int)
    df['LocationID'] = pd.to_numeric(df['LocationID'], errors='coerce').fillna(1).astype(int)


    df['Occupancy'] = df.apply(lambda row: min(row['Occupancy'], row['Capacity']), axis=1)


    df = df.sort_values('Timestamp').reset_index(drop=True)

    print(f"\nFinal dataset info:")
    print(f"Rows: {len(df)}")
    print(f"Date range: {df['Timestamp'].min()} to {df['Timestamp'].max()}")
    print(f"Unique locations: {df['LocationID'].nunique()}")
    print(f"Occupancy range: {df['Occupancy'].min()} to {df['Occupancy'].max()}")
    print(f"Capacity range: {df['Capacity'].min()} to {df['Capacity'].max()}")

    return df

# Model 1 Implementation
def implement_model1_pricing(df):
    """
    Implement Model 1 pricing with the processed data
    """
    # Model 1 Parameters
    BASE_PRICE = 10.0
    ALPHA = 5.0
    MIN_PRICE = 5.0
    MAX_PRICE = 25.0

    # Save data for streaming
    df[["Timestamp", "Occupancy", "Capacity", "LocationID"]].to_csv("parking_stream_model1.csv", index=False)

    # Define schema for Pathway
    class ParkingSchemaModel1(pw.Schema):
        Timestamp: str
        Occupancy: int
        Capacity: int
        LocationID: int

    # Load data as stream
    try:
        data = pw.demo.replay_csv("parking_stream_model1.csv", schema=ParkingSchemaModel1, input_rate=50)

        # Process the data
        fmt = "%Y-%m-%d %H:%M:%S"
        data_with_time = data.with_columns(
            t=data.Timestamp.dt.strptime(fmt),
            occupancy_rate=data.Occupancy / data.Capacity,
            location_key=pw.cast(str, data.LocationID)
        )

        # Implement pricing with tumbling windows
        pricing_stream = (
            data_with_time.windowby(
                pw.this.t,
                instance=pw.this.location_key,
                window=pw.temporal.tumbling(timedelta(minutes=30)),
                behavior=pw.temporal.exactly_once_behavior()
            )
            .reduce(
                location_id=pw.this._pw_instance,
                window_start=pw.this._pw_window_start,
                window_end=pw.this._pw_window_end,
                latest_occupancy_rate=pw.reducers.argmax(pw.this.t, pw.this.occupancy_rate),
                avg_occupancy_rate=pw.reducers.avg(pw.this.occupancy_rate),
                max_occupancy_rate=pw.reducers.max(pw.this.occupancy_rate),
                min_occupancy_rate=pw.reducers.min(pw.this.occupancy_rate),
                capacity=pw.reducers.max(pw.this.Capacity),
                observation_count=pw.reducers.count()
            )
            .with_columns(
                # Model 1: Linear pricing
                raw_price=BASE_PRICE + (ALPHA * pw.this.latest_occupancy_rate),
                price_adjustment=ALPHA * pw.this.latest_occupancy_rate
            )
            .with_columns(

                final_price=pw.apply(
                    lambda x: max(MIN_PRICE, min(MAX_PRICE, x)),
                    pw.this.raw_price
                ),
                price_category=pw.apply(
                    lambda x: "Low" if x <= 8 else "Medium" if x <= 15 else "High",
                    pw.this.raw_price
                )
            )
        )

        print("Model 1 pricing stream created successfully!")
        return pricing_stream

    except Exception as e:
        print(f"Error creating pricing stream: {e}")
        return None

def analyze_pricing_behavior(df):
    """
    Analyze how Model 1 behaves with the actual data
    """
    BASE_PRICE = 10.0
    ALPHA = 5.0
    MIN_PRICE = 5.0
    MAX_PRICE = 25.0


    df['OccupancyRate'] = df['Occupancy'] / df['Capacity']


    df['RawPrice'] = BASE_PRICE + (ALPHA * df['OccupancyRate'])
    df['FinalPrice'] = df['RawPrice'].clip(MIN_PRICE, MAX_PRICE)


    print("\n" + "="*60)
    print("MODEL 1 PRICING ANALYSIS")
    print("="*60)

    location_analysis = df.groupby('LocationID').agg({
        'OccupancyRate': ['mean', 'std', 'min', 'max'],
        'FinalPrice': ['mean', 'std', 'min', 'max'],
        'Occupancy': 'mean',
        'Capacity': 'mean'
    }).round(2)

    print("\nPricing by Location:")
    print(location_analysis)


    print(f"\nOverall Statistics:")
    print(f"Average Occupancy Rate: {df['OccupancyRate'].mean():.2%}")
    print(f"Average Price: ${df['FinalPrice'].mean():.2f}")
    print(f"Price Range: ${df['FinalPrice'].min():.2f} - ${df['FinalPrice'].max():.2f}")
    print(f"Price Volatility (std): ${df['FinalPrice'].std():.2f}")


    print(f"\nPrice Distribution:")
    price_bins = [0, 8, 15, 25, 100]
    price_labels = ['Low (<$8)', 'Medium ($8-15)', 'High ($15-25)', 'Very High (>$25)']
    df['PriceCategory'] = pd.cut(df['FinalPrice'], bins=price_bins, labels=price_labels)
    print(df['PriceCategory'].value_counts())

    return df

def simulate_model1_scenarios():
    """
    Simulate different occupancy scenarios
    """
    BASE_PRICE = 10.0
    ALPHA = 5.0
    MIN_PRICE = 5.0
    MAX_PRICE = 25.0

    print("\n" + "="*60)
    print("MODEL 1 SCENARIO SIMULATION")
    print("="*60)
    print("Formula: Price = Base_Price + α × (Occupancy/Capacity)")
    print(f"Parameters: Base=${BASE_PRICE}, α={ALPHA}, Range=[${MIN_PRICE}-${MAX_PRICE}]")
    print("-"*60)

    scenarios = [
        ("Empty lot", 0.0),
        ("Very low occupancy", 0.1),
        ("Low occupancy", 0.3),
        ("Medium occupancy", 0.5),
        ("High occupancy", 0.7),
        ("Very high occupancy", 0.9),
        ("Full lot", 1.0)
    ]

    print(f"{'Scenario':<20} {'Occupancy':<12} {'Raw Price':<10} {'Final Price':<12}")
    print("-"*60)

    for scenario_name, occupancy_rate in scenarios:
        raw_price = BASE_PRICE + (ALPHA * occupancy_rate)
        final_price = max(MIN_PRICE, min(MAX_PRICE, raw_price))

        print(f"{scenario_name:<20} {occupancy_rate:>8.1%} ${raw_price:>8.2f} ${final_price:>10.2f}")


if __name__ == "__main__":
    print("Starting Model 1 Implementation...")


    dataset_path = "/content/dataset.csv"
    df = robust_data_preprocessing(dataset_path)

    if df is not None:

        df_with_pricing = analyze_pricing_behavior(df)


        simulate_model1_scenarios()


        print("\n" + "="*60)

        print("="*60)


        pricing_stream = implement_model1_pricing(df)
        if pricing_stream:
          pw.run()

    else:
        print("Failed to process data. Please check your dataset.")
```
